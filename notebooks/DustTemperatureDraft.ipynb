```python
import xarray as xr
import yaml
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import sys
import os
import numpy as np
sys.path.append(os.path.dirname('../workflow'))
from workflow.scripts.plotting_tools import create_facet_plot,get_model_colordict, global_map
from workflow.scripts.utils import convert_lon_to_180, model_levels_to_pressure_levels, global_avg
```

```python
ds_exp = xr.open_dataset('../results/piClim-2xdust/ta/ta_piClim-2xdust_NorESM2-LM_Amon.nc')
ds_ctrl = xr.open_dataset('../results/piClim-control/ta/ta_piClim-control_NorESM2-LM_Amon.nc')
```

```python
with open('../config/config.yaml', 'r') as f:
    conf_dict = yaml.safe_load(f)
```

```python
nAF = conf_dict['dust_interest_regions']['North Africa']
ea = conf_dict['dust_source_regions']['East Asia']
```

```python
plevels=np.array([100000.,  92500., 87500. , 85000., 80000. ,75000., 70000.,  60000.,  50000.,  40000., 30000.,  20000.,    10000.,  5000., 500.,    100.])
```

```python
ds_exp = convert_lon_to_180(ds_exp)
ds_exp = ds_exp.sel(lon=slice(nAF['lonmin'],nAF['lonmax']),lat=slice(nAF['latmin'],nAF['latmax']))
ds_exp = model_levels_to_pressure_levels(ds_exp, plevels=plevels)
ds_exp = global_avg(ds_exp).compute()

ds_ctrl = convert_lon_to_180(ds_ctrl)
ds_ctrl= ds_ctrl.sel(lon=slice(nAF['lonmin'],nAF['lonmax']),lat=slice(nAF['latmin'],nAF['latmax']))
ds_ctrl = model_levels_to_pressure_levels(ds_ctrl,plevels=plevels)
ds_ctrl = global_avg(ds_ctrl).compute()
```

```python
fig,ax = plt.subplots(figsize=(8,7))
ds_ctrl.mean(dim='time')['ta'].plot(ax=ax,y='plev')
ds_exp.mean(dim='time')['ta'].plot(ax=ax, y='plev')
ax.set_ylim(100000, 50000)
ax.set_xlim(280,300)
```

```python
fig,ax = plt.subplots(figsize=(8,7))
(ds_exp.mean(dim='time')['ta']-ds_ctrl.mean(dim='time')['ta']).plot(ax=ax, y='plev')
ax.set_ylim(100000, 50000)
```
